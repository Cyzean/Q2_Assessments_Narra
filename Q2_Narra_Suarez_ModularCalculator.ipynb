```python
# Makes users input the values needed
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

# Makes the user choose an operation to use
operation = str(input("What operation? A = addition, B = subtraction, C = Multiplication, D = division: "))

# Is the function that calculates the sum between two numbers
def add(num1, num2):
  sum = num1 + num2
  return sum

# Is the function that calculates the difference between two numbers
def subtract(num1, num2):
  difference = num1 - num2
  return difference

# Is the function that calculates the product of two numbers
def multiply(num1, num2):
  product = num1 * num2
  return product

# Is the function that calculates the quotient of two numbers and errors when a value is equal to 0
def divide(num1, num2):
  if num1 == 0:
    num1 = int(input("First number must not be zero: "))
  if num2 == 0:
    num2 = int(input("Second number must not be zero: "))
  quotient = num1 / num2
  return quotient

# Calls the function "add" if "A" is chosen
if operation == "A":
  print(add(num1, num2))
# Calls the function "subtract" if "B" is chosen
elif operation == "B":
  print(subtract(num1, num2))
# Calls the function "multiply" if "C" is chosen
elif operation == "C":
  print(multiply(num1, num2))
# Calls the function "divide" if "D" is chosen
elif operation == "D":
  print(divide(num1, num2))
```
